```python
click=df.groupby("channel")["clicks"].sum()
click.plot(kind="bar", title="Total Clicks by Channel")
plt.xlabel("Channel")
plt.ylabel("Total Clicks")
plt.xticks(rotation=45)
plt.show()
```

```python
revenue=df.groupby("channel")["revenue"].sum()
revenue.plot(kind="bar", title="Total Revenue by Channel")
plt.xlabel("Channel")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
spend=df.groupby("channel")["spend"].sum()
spend.plot(kind="bar", title="Total Spend by Channel")
plt.xlabel("Channel")
plt.ylabel("Total Spend")
plt.xticks(rotation=45)
plt.show()
```

```python
conversions=df.groupby("channel")["conversions"].sum()
conversions.plot(kind="bar", title="Total Conversions by Channel")
plt.xlabel("Channel")
plt.ylabel("Conversions")
plt.xticks(rotation=45)
plt.show()
```

```python
#Performance Metrics
total_clicks=df["clicks"].sum()
total_conversions=df["conversions"].sum()
total_spend=df["spend"].sum()
total_revenue=df["revenue"].sum()
print("Total Clicks:",total_clicks)
print("Total Conversions:",total_conversions)
print("Total Spend:",total_spend)
print("Total Revenue:",total_revenue)

if total_clicks!= 0:
    print("Conversion Rate (%):",total_conversions/total_clicks*100)

if total_conversions !=0:
    print("Cost per Conversion:",total_spend/total_conversions)

if total_spend != 0:
    print("ROAS:",total_revenue/total_spend)
```

```python
channel_analysis=df.groupby("channel").agg({
    "clicks":"sum",
    "conversions":"sum",
    "spend":"sum",
    "revenue":"sum"
})

channel_analysis["conversion_rate"]=(
    channel_analysis["conversions"]/channel_analysis["clicks"]*100
)

channel_analysis["cost_per_conversion"]=(
    channel_analysis["spend"]/channel_analysis["conversions"].replace(0, np.nan)
)

channel_analysis["ROAS"]=(
    channel_analysis["revenue"]/
    channel_analysis["spend"].replace(0,np.nan)
)
channel_analysis.round(2)
```
